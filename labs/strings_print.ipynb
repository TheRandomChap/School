Denne notebook handler om at udskrive tekst på forskellige måder og arbejde videre med strenge i Python.

## Print af strenge på forskellige måder

Man kan printe strenge på flere måder i Python.

Tre almindelige metoder er:

- printe flere værdier med komma
- samle tekst med `+`
- bruge f-strings

### Eksempel: Tre måder at printe samme besked

F-strings er ofte den mest læsevenlige metode, når man skal indsætte variabler i tekst.

```python
navn = "Amina"
alder = 17

print("Hej", navn, "du er", alder, "år")
print("Hej " + navn + " du er " + str(alder) + " år")
print(f"Hej {navn}, du er {alder} år")
```

### Print en præsentation

Lav en f-string, der printer en kort præsentation med variablerne `navn`, `klasse` og `fag`.

```python
navn = "Noah"
klasse = "2.g"
fag = "Programmering B"
# indsæt din kode her
print(f"Hej {navn} lige nu har du {fag} i {klasse}")
```

### Print et regnestykke

Brug en f-string til at printe teksten `7 + 5 = 12` ved hjælp af variablerne herunder.

```python
a = 7
b = 5
# indsæt din kode her
print(f"{a} + {b} = {a+b}")
```

### Print med linjeskift

Brug `\n` til at printe navn, klasse og fag på tre forskellige linjer.

```python
navn = "Sara"
klasse = "2.g"
fag = "Programmering B"
# indsæt din kode her
print(f" Hej {navn} \n Lige nu her du {fag} \n Du går i {klasse}")
```

## Konkatenation

Konkatenation betyder, at man sætter strenge sammen.

I Python kan man bruge `+` til at sætte strenge sammen.

### Eksempel: Sæt fornavn og efternavn sammen

Bemærk, at man selv skal huske mellemrum.

```python
fornavn = "Ada"
efternavn = "Lovelace"
fulde_navn = fornavn + " " + efternavn
print(fulde_navn)
```

### Lav en hel sætning

Sæt ordene sammen, så programmet printer `Python er et programmeringssprog`.

```python
ord1 = "Python"
ord2 = "er"
ord3 = "et"
ord4 = "programmeringssprog"
# indsæt din kode her
sætning = ord1 + " " + ord2 + " " + ord3 + " " + ord4
print(sætning)
```

### Byg en mailadresse

Brug konkatenation til at bygge mailadressen `ida@skole.dk`.

```python
brugernavn = "ida"
domæne = "skole.dk"
# indsæt din kode her
adresse = brugernavn + "@" + domæne
print(adresse)
```

### Gentag en streng

Brug `*` til at printe strengen `ha` fire gange efter hinanden.

```python
tekst = "ha"
# indsæt din kode her

print(tekst*4)
```

## Længde og medlemskab

`len()` fortæller, hvor mange tegn der er i en streng.

Operatoren `in` kan bruges til at undersøge, om en tekst findes i en anden tekst.

### Eksempel: Længde og søgning

Her finder vi længden af en tekst og undersøger, om den indeholder ordet `data`.

```python
tekst = "data science"
print(len(tekst))

# metode 1
if "data" in tekst:
    print("Fundet!") 

# metode 2
print("Fundet!") if "data" in tekst else print("Nej, den var der ikke")
```

### Find tekstens længde

Udskriv længden af strengen herunder.

```python
tekst = "informatik"
# indsæt din kode her
print(len(tekst))
#print("Teksten er" len(tekst) " karektere langt;)") | virker ikke :(
```

### Undersøg om ordet findes

Undersøg med `in`, om ordet `kode` findes i teksten.

```python
tekst = "vi skriver kode i python"
# indsæt din kode her
if "kode" in tekst:
    print("Jeps, lige her!")
else:
    print("Nope")
```

### Tjek brugerens svar

Lav en simpel test, der printer `Korrekt`, hvis variablen `svar` indeholder ordet `python`.

```python
svar = "jeg kan godt lide python"
# indsæt din kode her
print("Fundet!") if "python" in svar else print()
```

## Nyttige string-metoder

Strings har indbyggede metoder. En metode kaldes med punktum.

Eksempler:

- `.lower()` laver teksten om til små bogstaver
- `.upper()` laver teksten om til store bogstaver
- `.strip()` fjerner mellemrum i starten og slutningen
- `.replace()` erstatter noget tekst med noget andet

### Eksempel: Ryd op i tekst

Her fjerner vi unødige mellemrum og laver teksten om til små bogstaver.

```python
tekst = "   PYTHON   "
ren_tekst = tekst.strip().lower()
print(ren_tekst)
```

### Gør teksten ensartet

Lav teksten om til små bogstaver og print resultatet.

```python
tekst = "ProGrAmMeRiNg"
Resultat1 = tekst.lower()
Resultat = tekst[0].upper() + tekst[1:].lower()
print(Resultat)
print(Resultat1)
# indsæt din kode her
```

### Fjern mellemrum

Brug `.strip()` til at fjerne mellemrum før og efter teksten.

```python
tekst = "   hej med dig   "
# indsæt din kode her
print(tekst.strip())
```

### Erstat et ord

Brug `.replace()` til at ændre `Java` til `Python`.

```python
tekst = "Jeg lærer Java"
# indsæt din kode her
text2 = tekst.replace("Java", "python")
print(text2)
# print(f " tekst.replace("java", "python")") | Virker ikke
```
