# Opgaver med strings i Python

Denne notebook handler om datatypen `string`, altså tekststrenge i Python.

I opgaverne skal du skrive kode i cellerne under hver opgave. Der er som regel allerede lavet en variabel, du kan arbejde videre med.

En string kan forstås som en liste af tegn. Hvert tegn har en plads i strengen, og pladserne kaldes indeks.

Eksempel:

```python
tekst = "Python"
print(tekst[0])  # P
print(tekst[1])  # y
```

Husk - første tegn har indeks 0. 

## Iteration af strenge

Når man itererer over en streng, gennemløber man tegnene et ad gangen.

Det kan gøres med en `for`-løkke.

### Eksempel: Print hvert tegn

Her gennemløber vi strengen `hej` og printer hvert tegn på sin egen linje.

`````python
tekst = "hej"

for tegn in tekst:
    print(tegn)
`````

### Udskriv bogstaverne under hinanden

Givet strengen `programmering`, udskriv hvert tegn på sin egen linje.

`````python
tekst = "programmering"
# indsæt din kode her
for tegn in tekst:
    print(tegn)
`````

### Tæl antallet af tegn

Brug en løkke til at tælle, hvor mange gange tegn teksten indeholder.

Tip: I stedet for at udskrive tegnet så læg 1 til `antal` hver gang løkken gennemløbes.


`````python
tekst = "Denne tekst er 28 tegn lang."
antal = 0
# indsæt din kode her
for tegn in tekst:
    antal = antal + 1

print(antal)
`````

### Tæl antal a'er

Givet strengen herunder, brug en løkke til at tælle, hvor mange gange bogstavet `a` forekommer.
Her skal du bruge en `if` statement og en tæller. Det ser fx sådan her ud:

```python
if a==b:
    antal = 1
````
Bemærk tre ting:

- at man bruger `==` (dobbelt lighedstegn) ved sammenligninger
- det afsluttende kolon
- indrykningen i næste linje



`````python
tekst = "banananalyse"
antal = 0

# NumA = 


for tegn in tekst:
   if tegn == "a":
     antal = antal + 1
print(antal)
`````

### Byg en ny streng uden mellemrum

Givet strengen herunder, brug en løkke til at lave en ny streng uden mellemrum.

`````python
tekst = "python er sjovt"
ny_tekst = ""
# indsæt din kode her
for tegn in tekst:
    if tegn != " ":
        ny_tekst = ny_tekst + tegn
print(ny_tekst)
`````

## Slicing med []

Slicing bruges til at tage en del af en streng.

Syntaksen er:

```python
streng[start:stop]
```

`start` er med, men `stop` er ikke med. Hvis vi skriver `tekst[0:4]`, får vi derfor tegnene med indeks `0`, `1`, `2` og `3`, men ikke tegnet med indeks `4`.

Man kan også skrive slicing med et tredje tal:

```python
streng[start:stop:trin]
```

`trin` bestemmer, hvor mange pladser Python hopper for hver gang. Et trin på `2` tager hvert andet tegn. Et trin på `-1` går baglæns gennem strengen.


### Eksempel: Tag de første tegn

Her tager vi de første fire tegn fra en streng.

`````python
tekst = "programmering"
print(tekst[0:4])
`````

### Negative indeks

Negative indeks tæller bagfra i strengen.

Det sidste tegn har indeks `-1`, det næstsidste har indeks `-2`, og sådan fortsætter det baglæns.

Det er praktisk, når man vil hente noget fra slutningen af en streng uden først at tælle alle tegnene.


`````python
tekst = "programmering"

print(tekst[-1])    # sidste tegn
print(tekst[-4:])   # de sidste fire tegn
print(tekst[-7:-4]) # fra indeks -7 til -4, men -4 er ikke med
`````

### Udeladte indeks

I en slice må man udelade `start` eller `stop`.

Hvis `start` udelades, starter Python fra begyndelsen af strengen. Hvis `stop` udelades, fortsætter Python helt til slutningen.

Det gør koden kortere, når man vil have begyndelsen eller slutningen af en streng.


`````python
tekst = "datavidenskab"

print(tekst[:5])  # fra starten til indeks 5, men 5 er ikke med
print(tekst[4:])  # fra indeks 4 og resten af strengen
print(tekst[:])   # hele strengen
`````

### Vend en streng om

Man kan vende en streng om med slicing ved at bruge trin `-1`.

Når trinnet er negativt, går Python baglæns. Med `tekst[::-1]` udelader vi både `start` og `stop`, så Python starter ved slutningen og fortsætter baglæns helt til begyndelsen.


`````python
tekst = "abcdef"
print(tekst[::-1])
`````

### Hent navnet

### Udskriv alle lige tal
 Tip: brug "trin" - det tredje parameter.

`````python
talliste = "012345678"
# indsæt din kode her
print(talliste[::2])

# better if not organised

list = ""
for tegn in talliste:
    if int(tegn) % 2 == 0:
        list = list + tegn
print(list)
`````

## Afsluttende miniopgave

Skriv et lille program, der arbejder med en valgfri tekst.

Programmet skal:

- tælle hvor mange tegn teksten indeholder ved hjælp af en løkke
- tælle hvor mange gange bogstavet `a` forekommer
- lave en ny tekst uden mellemrum
- udskrive de første fem tegn med slicing
- udskrive de sidste fem tegn med slicing
- udskrive teksten bagfra med slicing

`````python
tekst = "Programering er super sjovt, og meget lærerigt!"

antal_tegn = 0
antal_a = 0
uden_mellemrum = ""

for tegn in tekst:
    antal_tegn += 1

    if tegn == "a":
        antal_a += 1

    if tegn != " ":
        uden_mellemrum += tegn

print("Antal tegn:", antal_tegn)
print("Antal a'er:", antal_a)
print("Uden mellemrum:", uden_mellemrum)
print("De første fem tegn:", tekst[:5])
print("De sidste fem tegn:", tekst[-5:])
print("Teksten bagfra:", tekst[::-1])
`````
